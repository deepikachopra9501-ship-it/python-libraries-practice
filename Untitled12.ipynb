```python
import sqlite3

# Connect to database
conn = sqlite3.connect("student.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS studentS (
    roll_no INTEGER PRIMARY KEY,
    student_name TEXT,
    email TEXT,
    group_name TEXT
)
""")

# Insert multiple records
cursor.executemany("""
INSERT OR REPLACE INTO studentS
(roll_no, student_name, email, group_name)
VALUES (?, ?, ?, ?)
""", (
    (102, "Aman Kumar", "aman@chitkara.edu.in", "3g1"),
    (101, "Simran", "simran@chitkara.edu.in", "3g2"),
    (103, "Rahul Sharma", "rahul@chitkara.edu.in", "3g1"),
    (104, "Priya", "priya@chitkara.edu.in", "3g2")
))

# Save changes
conn.commit()

# Display records
cursor.execute("SELECT * FROM studentS")

for row in cursor.fetchall():
    print(row)

# Close connection
cursor.close()
conn.close()
```
