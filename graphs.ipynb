```python
#ECO
import numpy as np
import matplotlib.pyplot as plt

cover = np.linspace(0, 100, 501)
species_loss = np.maximum(0, 0.90 * cover - 2.81)

plt.figure(figsize=(8, 5.5))

plt.axvspan(0, 25, alpha=0.12, color="green",
            label="Monitor")
plt.axvspan(25, 55, alpha=0.12, color="orange",
            label="Intervene")
plt.axvspan(55, 100, alpha=0.10, color="red",
            label="Critical")

plt.plot(cover, species_loss, color="black",
         linewidth=2.2, label="Predicted species loss")

plt.axvline(25, linestyle="--", color="gray")
plt.axvline(55, linestyle="--", color="gray")

plt.xlabel("Mimosa pigra cover (%)")
plt.ylabel("Predicted native-species loss (%)")
plt.title("Predicted Native-Species Loss vs. Mimosa pigra Cover")

plt.xlim(0, 100)
plt.ylim(0, 90)
plt.grid(True, alpha=0.2)
plt.legend(frameon=False)
plt.tight_layout()

plt.savefig("Figure1_Eco.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure1_Eco.pdf", bbox_inches="tight")
plt.show()
```

```python
#FAUNA

import numpy as np
import matplotlib.pyplot as plt

area = np.linspace(536, 6604, 500)
nests = -0.01 * area**2 + 71.4 * area - 35410

peak_area = 71.4 / 0.02
peak_nests = -0.01 * peak_area**2 + 71.4 * peak_area - 35410

plt.figure(figsize=(8, 5.5))

plt.plot(area, nests, color="black", linewidth=2.5,
         label="Model-predicted nest loss")

plt.scatter(peak_area, peak_nests, color="red", zorder=3,
            label=f"Model peak: {peak_area:.0f} km²")

plt.axvline(peak_area, linestyle="--", alpha=0.6)

plt.xlabel("Floodplain area lost to Mimosa pigra (km²)")
plt.ylabel("Predicted magpie goose nests lost")
plt.title("Predicted Magpie Goose Nest Loss vs. Floodplain Area Lost")

plt.grid(True, alpha=0.2)
plt.legend(frameon=False)
plt.tight_layout()

plt.savefig("Figure2_Fauna.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure2_Fauna.pdf", bbox_inches="tight")
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

density = np.array([100, 500, 2000, 5000])
effort = 4.78 * density**(-0.93)

plt.figure(figsize=(8, 5.5))

plt.plot(density, effort, marker="o", color="black",
         linewidth=2, markersize=6)

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Infestation Density (plants per plot)")
plt.ylabel("Predicted Control Effort (days per plant)")
plt.title("Predicted Control Effort vs. Infestation Density")

plt.grid(True, which="both", alpha=0.2)
plt.tight_layout()

plt.savefig("Figure3_Ops.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure3_Ops.pdf", bbox_inches="tight")
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

years = np.array([0, 1, 2, 3, 4])

cover = np.array([15.0, 23.8, 37.8, 60.0, 95.2])
species_loss = np.array([10.7, 18.6, 31.2, 51.2, 82.9])

fig, axes = plt.subplots(1, 2, figsize=(11, 4.8))

axes[0].plot(years, cover, marker="o", linewidth=2)
axes[0].set_title("Projected Mimosa Cover")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("Cover (%)")
axes[0].set_xticks(years)
axes[0].set_ylim(0, 100)
axes[0].grid(True, alpha=0.25)

axes[1].plot(years, species_loss, marker="s", linewidth=2)
axes[1].set_title("Modelled Species Loss")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Predicted Species Loss (%)")
axes[1].set_xticks(years)
axes[1].set_ylim(0, 100)
axes[1].grid(True, alpha=0.25)

fig.suptitle("Four-Year Hypothetical Mimosa Invasion Scenario")
plt.tight_layout()

plt.savefig("Figure5_Projection.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure5_Projection.pdf", bbox_inches="tight")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

scenarios = ["Act Now", "Wait 3 Years"]

cost = [120900, 180700]
species_loss = [10.7, 51.2]

fig, axes = plt.subplots(1, 2, figsize=(10, 4.8))

axes[0].bar(scenarios, cost)
axes[0].set_title("Modelled Treatment Cost")
axes[0].set_ylabel("Historical Cost (AU$)")
axes[0].grid(axis="y", alpha=0.25)

for i, value in enumerate(cost):
    axes[0].text(i, value, f"{value:,}", ha="center", va="bottom")

axes[1].bar(scenarios, species_loss)
axes[1].set_title("Predicted Species Loss")
axes[1].set_ylabel("Species Loss (%)")
axes[1].set_ylim(0, 60)
axes[1].grid(axis="y", alpha=0.25)

for i, value in enumerate(species_loss):
    axes[1].text(i, value, f"{value}%", ha="center", va="bottom")

fig.suptitle("Modelled Comparison: Immediate Action vs. Three-Year Delay")
plt.tight_layout()

plt.savefig("Figure6_Action_vs_Delay.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure6_Action_vs_Delay.pdf", bbox_inches="tight")
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

doubling_time = np.linspace(1.4, 1.5, 100)

# EEDR for the paper's three-year delay scenario
# Species-loss ratio: 51.2 / 10.7
# Cost ratio: 180700 / 120900
species_loss_ratio = 51.2 / 10.7
cost_ratio = 180700 / 120900

eedr = np.full_like(
    doubling_time,
    species_loss_ratio / cost_ratio
)

plt.figure(figsize=(8, 5))

plt.plot(doubling_time, eedr, linewidth=2)
plt.xlabel("Assumed Invasion Doubling Time (years)")
plt.ylabel("EEDR")
plt.title("EEDR Sensitivity to Invasion Doubling Time")
plt.grid(True, alpha=0.25)

plt.tight_layout()
plt.savefig("Figure7_EEDR.png", dpi=300, bbox_inches="tight")
plt.savefig("Figure7_EEDR.pdf", bbox_inches="tight")
plt.show()

print("EEDR =", round(eedr[0], 2))
```

```python

import numpy as np
import matplotlib.pyplot as plt

def N(A):
    return -0.01 * A**2 + 71.4 * A - 35410

A = np.linspace(536.2, 6603.8, 800)

fig, ax = plt.subplots(figsize=(8, 5))

asc = A <= 3500

ax.plot(A[asc], N(A[asc]), "k", linewidth=2,
        label="Ascending branch")
ax.plot(A[~asc], N(A[~asc]), color="gray",
        linewidth=2, linestyle="--",
        label="Descending branch (quadratic artifact)")

ax.plot(3570, N(3570), "ro",
        label="Vertex: 3,570 km²")

ax.axvline(800, color="C0", linestyle=":",
           label="Documented extent: 800 km²")
ax.axvline(3500, color="C1", linestyle=":",
           label="Source plot limit: 3,500 km²")

ax.set_xlabel("Floodplain area lost to Mimosa pigra (km²)")
ax.set_ylabel("Predicted magpie goose nests lost")
ax.legend(fontsize=8)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.savefig("fig2_fauna.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

def E(D):
    return 4.78 * D**(-0.93)

D = np.logspace(np.log10(15), np.log10(5200), 200)
points = [100, 500, 2000, 5000]

plt.figure(figsize=(8, 5))
plt.loglog(D, E(D), "k", linewidth=2,
           label="E = 4.78D^-0.93")
plt.loglog(points, [E(d) for d in points], "ko",
           label="Table values")
plt.xlabel("Infestation density (plants per plot)")
plt.ylabel("Predicted effort (person-days per plant)")
plt.title("Ops Module")
plt.grid(True, which="both", alpha=0.25)
plt.legend()
plt.tight_layout()
plt.savefig("fig3_ops.png", dpi=300, bbox_inches="tight")
plt.show()
```
